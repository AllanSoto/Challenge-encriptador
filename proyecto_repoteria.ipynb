```python
import sqlite3
from faker import Faker
import random
from datetime import date

# Conectar a SQLite (en memoria, también podés usar archivo si querés persistencia)
conn = sqlite3.connect('cobranzas.db')
cursor = conn.cursor()

# Borrar tabla si existe
cursor.execute("DROP TABLE IF EXISTS clientes")

# Crear la tabla
cursor.execute("""
CREATE TABLE clientes (
  id INTEGER PRIMARY KEY AUTOINCREMENT,
  nombre TEXT,
  proyecto TEXT,
  categoria TEXT,
  subcategoria TEXT,
  mora_dias INTEGER,
  deuda REAL,
  recupero REAL,
  promesa_pago REAL,
  fecha_registro DATE,
  estado TEXT
)
""")

# Preparar generador
fake = Faker('es_ES')

proyectos = [f'Proyecto {i}' for i in range(1, 6)]
categorias = {
    'Telecomunicaciones': ['Claro', 'Movistar', 'Telecable'],
    'Bancos': ['Scotiabank', 'Daviviena']
}

años = [2022, 2023, 2024, 2025]
registros_por_año = 250

# Insertar 1000 registros
for año in años:
    for i in range(registros_por_año):
        mes = random.randint(1, 4) if año == 2025 else random.randint(1, 12)
        dia = random.randint(1, 28)
        fecha_registro = date(año, mes, dia)

        nombre = fake.name()
        proyecto = random.choice(proyectos)
        categoria = random.choice(list(categorias.keys()))
        subcategoria = random.choice(categorias[categoria])
        mora_dias = random.randint(1, 89) if i < int(0.20 * registros_por_año) else random.randint(181, 365)
        deuda = round(random.uniform(100.0, 5000.0), 2)
        recupero = round(random.uniform(0, deuda), 2)
        promesa_pago = round(random.uniform(0, deuda), 2)
        estado = 'Administrativa' if mora_dias < 90 else 'Incobrable'

        cursor.execute("""
            INSERT INTO clientes (nombre, proyecto, categoria, subcategoria, mora_dias, deuda, recupero, promesa_pago, fecha_registro, estado)
            VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?, ?)
        """, (nombre, proyecto, categoria, subcategoria, mora_dias, deuda, recupero, promesa_pago, fecha_registro, estado))

conn.commit()
print("✅ Base de datos SQLite con 1000 registros creada con éxito.")
```

```python
import sqlite3
import pandas as pd

# Conectar a la base de datos
conn = sqlite3.connect("recupero.db")

# Leer la tabla
clientes_df = pd.read_sql_query("SELECT * FROM clientes", conn)

# Convertir fecha a datetime
clientes_df["fecha"] = pd.to_datetime(clientes_df["fecha"])
clientes_df["anio"] = clientes_df["fecha"].dt.year
clientes_df["mes"] = clientes_df["fecha"].dt.month
clientes_df["dia"] = clientes_df["fecha"].dt.day

# Agrupaciones necesarias
recupero_por_anio = clientes_df.groupby("anio")["recupero"].sum().reset_index()
promesas_por_anio = clientes_df.groupby("anio")["promesa"].sum().reset_index()

recupero_vs_promesas_mes = clientes_df.groupby(["anio", "mes"])[["recupero", "promesa"]].sum().reset_index()
recupero_vs_promesas_dia = clientes_df.groupby(["anio", "mes", "dia"])[["recupero", "promesa"]].sum().reset_index()

# Crear una columna booleana para el alcance
clientes_df["con_contacto"] = (clientes_df["recupero"] > 0) | (clientes_df["promesa"] > 0)

# Agrupar y calcular el alcance como porcentaje
alcance_base = clientes_df.groupby("anio").agg(
    clientes_totales=("id", "count"),
    clientes_contactados=("con_contacto", "sum")
).reset_index()

alcance_base["alcance_base"] = (
    alcance_base["clientes_contactados"] / alcance_base["clientes_totales"] * 100
)

```

```python
import sqlite3
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
from dash import Dash, dcc, html
import dash_bootstrap_components as dbc
from pyngrok import ngrok

# ------------------------------------------------------------------------------
# 1. Cargar los datos desde SQLite (asegúrate de que 'recupero.db' y la tabla 'clientes' existan)
# ------------------------------------------------------------------------------
conn = sqlite3.connect("recupero.db")
clientes_df = pd.read_sql_query("SELECT * FROM clientes", conn)
conn.close()

# Convertir 'fecha' a datetime y extraer año, mes y día
clientes_df["fecha"] = pd.to_datetime(clientes_df["fecha"])
clientes_df["anio"] = clientes_df["fecha"].dt.year
clientes_df["mes"] = clientes_df["fecha"].dt.month
clientes_df["dia"] = clientes_df["fecha"].dt.day

# ------------------------------------------------------------------------------
# 2. Calcular las métricas requeridas
# ------------------------------------------------------------------------------
# Total recuperado por año
recupero_por_anio = clientes_df.groupby("anio")["recupero"].sum().reset_index()
# Total promesas por año
promesas_por_anio = clientes_df.groupby("anio")["promesa"].sum().reset_index()

# Para la gráfica de recupero, creamos un DataFrame que muestre, para cada año,
# lo recuperado y la diferencia pendiente (total promesas - total recuperado)
df_ratio = pd.merge(recupero_por_anio, promesas_por_anio, on="anio")
df_ratio["Pendiente"] = df_ratio["promesa"] - df_ratio["recupero"]

# Transformamos a formato largo, para usarlo en un gráfico Sunburst (tipo donut)
df_long = pd.melt(
    df_ratio,
    id_vars=["anio"],
    value_vars=["recupero", "Pendiente"],
    var_name="Tipo",
    value_name="Monto"
)

# ------------------------------------------------------------------------------
# 3. Crear Gráficos con Plotly usando tema oscuro y paleta de tonos menta
# ------------------------------------------------------------------------------
# Definir paleta en tonos menta
menta_palette = ["#98FF98", "#8AE79C", "#73D389", "#5BCB75", "#44B160", "#2DA64C"]

# Gráfico Donut (Sunburst) para "Recupero Total por Año"
# Este gráfico mostrará, para cada año, la proporción de lo Recuperado y lo Pendiente (Promesas - Recupero)
fig_recupero_donut = px.sunburst(
    df_long,
    path=["anio", "Tipo"],
    values="Monto",
    color="anio",
    color_discrete_sequence=menta_palette,
    title="Recupero vs. Pendiente (de Promesas) por Año",
    template="plotly_dark"
)
# Opcional: actualizar el texto
fig_recupero_donut.update_traces(textinfo="label+percent entry")

# Gráfico de barras para "Promesas de Pago por Año"
fig_promesas = px.bar(
    promesas_por_anio,
    x="anio",
    y="promesa",
    title="Promesas de Pago por Año",
    labels={"promesa": "Monto en Promesas (₡)", "anio": "Año"},
    color="anio",
    color_discrete_sequence=menta_palette,
    template="plotly_dark"
)

# Conservamos otros gráficos (porcentaje de recupero, alcance, evolución mensual, etc.) si deseas
# Por ejemplo, un gráfico de línea para el porcentaje de recupero
porcentaje_recupero = clientes_df.groupby("anio").apply(
    lambda x: x["recupero"].sum() / x["deuda"].sum() * 100
).reset_index(name="porcentaje_recupero")

fig_porcentaje = px.line(
    porcentaje_recupero,
    x="anio",
    y="porcentaje_recupero",
    title="Porcentaje de Recupero por Año",
    template="plotly_dark",
    markers=True
)

# Tarjetas de KPI para el dashboard (utilizamos los totales calculados)
tarjeta_recupero = dbc.Card(
    dbc.CardBody([
        html.H4(f"₡{recupero_por_anio['recupero'].sum():,.2f}", className="card-title"),
        html.P("Total Recuperado", className="card-text")
    ]),
    color="info", inverse=True
)

tarjeta_promesas = dbc.Card(
    dbc.CardBody([
        html.H4(f"₡{promesas_por_anio['promesa'].sum():,.2f}", className="card-title"),
        html.P("Total Promesas de Pago", className="card-text")
    ]),
    color="secondary", inverse=True
)

tarjeta_porcentaje = dbc.Card(
    dbc.CardBody([
        html.H4(f"{porcentaje_recupero['porcentaje_recupero'].mean():.2f}%", className="card-title"),
        html.P("Porcentaje de Recupero", className="card-text")
    ]),
    color="primary", inverse=True
)

# ------------------------------------------------------------------------------
# 4. Crear el Dashboard con Dash y tema DARKLY
# ------------------------------------------------------------------------------
app = Dash(__name__, external_stylesheets=[dbc.themes.DARKLY])

app.layout = dbc.Container(
    fluid=True,
    children=[
        # Encabezado
        dbc.Row(
            dbc.Col(html.H1("Dashboard de Cobranzas", className="text-center text-light my-4")),
            className="mb-4"
        ),

        # Tarjetas KPI
        dbc.Row([
            dbc.Col(tarjeta_recupero, md=4),
            dbc.Col(tarjeta_promesas, md=4),
            dbc.Col(tarjeta_porcentaje, md=4)
        ], className="mb-4"),

        # Gráficos
        dbc.Row([
            dbc.Col(dcc.Graph(figure=fig_recupero_donut), md=6),
            dbc.Col(dcc.Graph(figure=fig_promesas), md=6)
        ], className="mb-4"),

        dbc.Row([
            dbc.Col(dcc.Graph(figure=fig_porcentaje), md=12)
        ], className="mb-4")
    ]
)

# ------------------------------------------------------------------------------
# 5. Exponer y ejecutar la aplicación en Colab usando ngrok
# ------------------------------------------------------------------------------
from pyngrok import ngrok

# Cierra túneles previos para evitar el límite
ngrok.kill()

public_url = ngrok.connect(addr="8050")
print("Tu dashboard está disponible en:", public_url)

if __name__ == '__main__':
    app.run(port=8050, debug=True)



```

```python
# DASHBOARD RESUELVA CR 2025

import pandas as pd
import dash
import dash_bootstrap_components as dbc
from dash import dcc, html, Input, Output
import plotly.express as px
from google.colab import drive

# Montar Google Drive
drive.mount('/content/drive')

# Ruta del archivo CSV
csv_path = '/content/drive/MyDrive/Colab Notebooks/Libro1.csv'

# Cargar datos y limpiar
columnas = ['id', 'nombre', 'proyecto', 'categoria', 'subcategoria', 'mora_dias', 'deuda',
            'recupero', 'promesa_pago', 'pendiente de pago', 'no recuperado en promesa',
            'fecha_registro', 'estado']

df = pd.read_csv(csv_path, sep=';', names=columnas, header=0)

# Limpieza de columnas numéricas
df['deuda'] = pd.to_numeric(df['deuda'], errors='coerce')
df['recupero'] = pd.to_numeric(df['recupero'], errors='coerce')
df['promesa_pago'] = pd.to_numeric(df['promesa_pago'], errors='coerce')

# Convertir fecha y desglosar

# Detectar el formato de fecha correcto automáticamente
try:
    df['fecha_registro'] = pd.to_datetime(df['fecha_registro'], dayfirst=True, errors='coerce')
except:
    df['fecha_registro'] = pd.to_datetime(df['fecha_registro'], errors='coerce')

# Extraer partes de la fecha
df['año'] = df['fecha_registro'].dt.year
df['mes'] = df['fecha_registro'].dt.month
df['dia'] = df['fecha_registro'].dt.day

# Filtrado 2025 hasta abril
df_2025 = df[(df['año'] == 2025) & (df['mes'] <= 4)]

# KPIs de 2025
recupero_2025 = df_2025['recupero'].sum()
promesas_2025 = df_2025['promesa_pago'].sum()

# Tarjetas
kpi_style = {'backgroundColor': '#1e1e1e', 'color': 'white', 'padding': '20px', 'borderRadius': '15px'}

tarjeta_recupero = dbc.Card([
    dbc.CardBody([
        html.H4("TOTAL RECUPERADO 2025", className="card-title text-success"),
        html.H2(f"\u20a11{recupero_2025:,.0f}", className="card-text")
    ])
], style=kpi_style)

tarjeta_promesas = dbc.Card([
    dbc.CardBody([
        html.H4("TOTAL PROMESAS 2025", className="card-title text-primary"),
        html.H2(f"\u20a11{promesas_2025:,.0f}", className="card-text")
    ])
], style=kpi_style)

porcentaje = (recupero_2025 / promesas_2025 * 100) if promesas_2025 > 0 else 0

tarjeta_porcentaje = dbc.Card([
    dbc.CardBody([
        html.H4("% RECUPERADO 2025", className="card-title text-info"),
        html.H2(f"{porcentaje:.2f}%", className="card-text")
    ])
], style=kpi_style)

# GRAFICAS

# Donut por subcategoría 2025
fig_recupero_donut = px.pie(df_2025, names='subcategoria', values='recupero', title='Recupero por Subcategoría 2025')
fig_recupero_donut.update_layout(paper_bgcolor='#1e1e1e', plot_bgcolor='#1e1e1e', font_color='white')

# Promesas por año (2022 a 2025)
df_multi = df[df['año'].isin([2022, 2023, 2024, 2025]) & (df['mes'] <= 4)]
promesas_año = df_multi.groupby('año')[['promesa_pago', 'recupero']].sum().reset_index()
fig_promesas = px.bar(promesas_año, x='año', y=['promesa_pago', 'recupero'], barmode='group', title='Promesas vs Recupero por Año')
fig_promesas.update_layout(paper_bgcolor='#1e1e1e', plot_bgcolor='#1e1e1e', font_color='white')

# Porcentaje de recupero diario
df_diario = df_2025.groupby('fecha_registro')[['recupero']].sum().reset_index()
fig_porcentaje = px.line(df_diario, x='fecha_registro', y='recupero', title='Recupero Diario en 2025')
fig_porcentaje.update_traces(mode="lines+markers")
fig_porcentaje.update_layout(paper_bgcolor='#1e1e1e', plot_bgcolor='#1e1e1e', font_color='white')

# Recupero vs Promesa por mes 2025
mensual = df_2025.groupby('mes')[['recupero', 'promesa_pago']].sum().reset_index()
fig_mes = px.bar(mensual, x='mes', y=['recupero', 'promesa_pago'], barmode='group', title='Recupero vs Promesa por Mes 2025')
fig_mes.update_layout(paper_bgcolor='#1e1e1e', plot_bgcolor='#1e1e1e', font_color='white')

# Recupero vs Promesa por Año
anual = df[df['año'] <= 2025].groupby('año')[['recupero', 'promesa_pago']].sum().reset_index()
fig_anual = px.bar(anual, x='año', y=['recupero', 'promesa_pago'], barmode='group', title='Recupero vs Promesa por Año')
fig_anual.update_layout(paper_bgcolor='#1e1e1e', plot_bgcolor='#1e1e1e', font_color='white')

# Avance en abril 2025
df_abril = df_2025[df_2025['mes'] == 4].groupby('dia')[['recupero', 'promesa_pago']].sum().reset_index()
fig_abril = px.line(df_abril, x='dia', y=['recupero', 'promesa_pago'], markers=True, title='Avance Abril 2025')
fig_abril.update_layout(paper_bgcolor='#1e1e1e', plot_bgcolor='#1e1e1e', font_color='white')

# Subcategoría recuperada
subcat_2025 = df_2025.groupby('subcategoria')['recupero'].sum().reset_index().sort_values(by='recupero', ascending=False)
fig_subcat = px.bar(subcat_2025, x='subcategoria', y='recupero', title='Recuperado por Subcategoría 2025')
fig_subcat.update_layout(paper_bgcolor='#1e1e1e', plot_bgcolor='#1e1e1e', font_color='white')

# Crear App con tema DARKLY
app = dash.Dash(__name__, external_stylesheets=[dbc.themes.DARKLY])

app.layout = dbc.Container(
    fluid=True,
    children=[
        dbc.Row(dbc.Col(html.H1("RESUELVA CR 2025", className="text-center text-light my-4"))),

        dbc.Row([
            dbc.Col(tarjeta_recupero, md=4),
            dbc.Col(tarjeta_promesas, md=4),
            dbc.Col(tarjeta_porcentaje, md=4)
        ], className="mb-4"),

        dbc.Row([
            dbc.Col(dcc.Graph(figure=fig_recupero_donut), md=6),
            dbc.Col(dcc.Graph(figure=fig_promesas), md=6)
        ], className="mb-4"),

        dbc.Row([
            dbc.Col(dcc.Graph(figure=fig_porcentaje), md=6),
            dbc.Col(dcc.Graph(figure=fig_mes), md=6)
        ], className="mb-4"),

        dbc.Row([
            dbc.Col(dcc.Graph(figure=fig_anual), md=6),
            dbc.Col(dcc.Graph(figure=fig_abril), md=6)
        ], className="mb-4"),

        dbc.Row([
            dbc.Col(dcc.Graph(figure=fig_subcat), md=12)
        ])
    ]
)

# Ejecutar app (en Colab cambiar a .run(mode='inline') si necesario)
app.run(mode='inline')



```
